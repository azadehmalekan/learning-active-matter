```python
from pathlib import Path
import numpy as np
from scipy.io import savemat

# Project root
project_root = Path.cwd().parent

# Data folder
data_folder = project_root / "data"

# Find all NPZ files
npz_files = sorted(data_folder.glob("run_*.npz"))

print("Found", len(npz_files), "NPZ files")

for npz_file in npz_files:

    # Load NPZ
    data = np.load(npz_file)

    # Convert all variables automatically
    mat_data = {
        key: data[key]
        for key in data.files
    }

    # Output MAT filename
    mat_file = npz_file.with_suffix(".mat")

    # Save
    savemat(mat_file, mat_data)

    print(f"Converted: {npz_file.name} -> {mat_file.name}")

print("All files converted.")
```
